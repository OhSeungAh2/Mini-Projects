### 02. 데이터 기본 정보 확인

```python
# 첫 번째 엑셀 파일의 교통량 데이터 시트
df = pd.read_excel(excel_files[0], sheet_name="2026년 01월")

# 데이터 크기
print("데이터 Shape:", df.shape)

# 컬럼명
print("\n컬럼 목록:")
print(df.columns.tolist())

# 상위 5행
display(df.head())
```

### 03. D-09 유입 데이터 확인

```python
# D-09 유입 방향 데이터만 추출
df_d09 = df[
    (df["지점번호"] == "D-09") &
    (df["방향"] == "유입")
].copy()

print("D-09 유입 데이터 Shape:", df_d09.shape)

print("\n관측 지점:", df_d09["지점명"].unique())
print("방향:", df_d09["방향"].unique())
print("구간:", df_d09["구분"].unique())

display(df_d09.head())
```

### 04. 데이터 병합

```python
df_list = []

for month, file in enumerate(excel_files, start=1):
    # 해당 월 교통량 데이터 불러오기
    sheet_name = f"2026년 {month:02d}월"
    monthly_df = pd.read_excel(file, sheet_name=sheet_name)

    # D-09 유입 방향 데이터만 추출
    monthly_d09 = monthly_df[
        (monthly_df["지점번호"] == "D-09") &
        (monthly_df["방향"] == "유입")
    ].copy()

    df_list.append(monthly_d09)

    print(f"{month}월: {len(monthly_d09)}일")

# DataFrame 합치기
df_traffic = pd.concat(df_list, ignore_index=True)

df_traffic = df_traffic.sort_values("일자").reset_index(drop=True)

print("\n전체 데이터 Shape:", df_traffic.shape)
print("시작 날짜:", df_traffic["일자"].min())
print("종료 날짜:", df_traffic["일자"].max())

display(df_traffic.head())
```

### 05. 데이터 전처리

```python
# 0시~23시 컬럼 생성
hour_cols = [f"{h}시" for h in range(24)]

# 시간별 컬럼 행 변환 (Wide → Long)
df_hourly = df_traffic.melt(
    id_vars=["일자"],
    value_vars=hour_cols,
    var_name="hour",
    value_name="traffic_volume"
)

# 시간 컬럼 정수 변환 (13시 → 13)
df_hourly["hour"] = (
    df_hourly["hour"]
    .str.replace("시", "", regex=False)
    .astype(int)
)

# 날짜 datetime 자료로 변환
df_hourly["date"] = pd.to_datetime(
    df_hourly["일자"].astype(str),
    format="%Y%m%d"
)

# 날짜, 시간 합쳐 실제 관측 시각 생성
df_hourly["datetime"] = (
    df_hourly["date"] +
    pd.to_timedelta(df_hourly["hour"], unit="h")
)

# 시간 순서대로 정렬
df_hourly = (
    df_hourly
    .sort_values("datetime")
    .reset_index(drop=True)
)

# 필요한 컬럼만
df_hourly = df_hourly[
    ["datetime", "traffic_volume"]
]

print("시간별 데이터 Shape:", df_hourly.shape)
print("시작 시각:", df_hourly["datetime"].min())
print("종료 시각:", df_hourly["datetime"].max())

display(df_hourly.head(10))
```
